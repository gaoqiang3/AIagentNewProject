```python
from pydantic_ai import Agent, RunContext

from dataclasses import dataclass
import os

api_key = os.getenv("OPENAI_API_KEY") or ""


@dataclass

class MyDeps:

    api_key: str


# 1. 创建 Agent 时指定 deps_type

agent = Agent('openai:qwen-plus', deps_type=MyDeps)


# 2. 工具函数的第一个参数必须是 ctx: RunContext[MyDeps]

@agent.tool
def get_weather(ctx: RunContext[MyDeps], city: str) -> str:

    """获取指定城市的实时天气。"""

# 现在你可以通过 ctx.deps 访问 api_key

    print(f"正在使用密钥 {ctx.deps.api_key} 查询 {city}")

    return f"{city}的天气是晴天"

```

```python
my_config = MyDeps(api_key=api_key)
# 将你的 main 函数改为异步定义
async def main():
    # 注意：这里改用 run() 而不是 run_sync()，并且前面加 await
    result = await agent.run(
        "公民有权通过分布式议会发起对算法代码的什么要求？",
        deps=my_config
    )
    print(result)

# 在 Jupyter/IPython 中直接 await 它
await main()
```
